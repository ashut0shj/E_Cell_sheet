```python
from __future__ import print_function

from googleapiclient.discovery import build

from google.oauth2 import service_account
import pandas as pd 


SCOPES = ['https://www.googleapis.com/auth/spreadsheets']
SERVICE_ACCOUNT_FILE = '/content/drive/MyDrive/Colab Notebooks/sheets/keys.json'

creds=None
creds = service_account.Credentials.from_service_account_file(
        SERVICE_ACCOUNT_FILE, scopes=SCOPES)
```

```python
#mail merge 
SAMPLE_RANGE_NAME_mm = "Sheet1!A1:C"
SAMPLE_SPREADSHEET_ID_mm = '1JS8r7GOYTbrYvhW3VJDfa_a18TiQypPnkErP078-QAo'


service = build('sheets', 'v4', credentials=creds)

# Call the Sheets API
sheet = service.spreadsheets()
result = sheet.values().get(spreadsheetId=SAMPLE_SPREADSHEET_ID_mm,
                            range=SAMPLE_RANGE_NAME_mm).execute()
valuesmm = result.get('values', [])

#extract data from mail merge sheet
datamm=[]
for i in valuesmm:
    try:
    
        datamm.append(i[1])
    except:
        pass
print(pd.DataFrame(datamm))
```
